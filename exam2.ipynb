```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['income_binary'] = df['income'].apply(lambda x: 1 if x == '>50K' else 0)
```

```python
df.columns
```

```python
df['age'].unique()
```

```python
df['age_group'] = df['age'].apply(lambda x: 'Senior' if x >= 60 else ('Adult' if x < 30 else 'Youth'))
```

```python
df.columns
```

```python
df.rename(columns={'educational-num': 'educational_num'}, inplace=True)
```

```python
df.head()

```

```python
df['workclass'] = df['workclass'].replace('?', 'Unknown')
df['occupation'] = df['occupation'].replace('?', 'Unknown')
df['workclass'].unique()
```

```python
df['occupation'].unique()
```

```python
average_age_by_workclass = df.groupby('workclass')['age'].mean()
print("Average Age by Workclass:")
print(average_age_by_workclass)
```

```python
total_capital_gain = df.groupby('gender')['capital-gain'].sum()
print("Total Capital Gain:")
print(total_capital_gain)
```

```python
count_marital_status = df.groupby('marital-status').size()
print("Count of Marital Status:")
print(count_marital_status)
```

```python
mean_hours_per_week = df.groupby('education')['hours-per-week'].mean()
print("Mean Hours Per Week:")
print(mean_hours_per_week)
```

```python
mean_fnlwgt = df.groupby('race')['fnlwgt'].mean()
print('Mean of fnlwgt')
print(mean_fnlwgt)

median_fnlwgt = df.groupby('race')['fnlwgt'].median()
print('Median of fnlwgt')
print(median_fnlwgt)
std_fnlwgt = df.groupby('race')['fnlwgt'].std()
print('STD of fnlwgt')
print(std_fnlwgt)
```

```python
df.head()
```

```python
count_country = df.groupby('native-country')['native-country'].count()
print("Count of Country:")
print(count_country)
```

```python


plt.figure(figsize=(12, 6))
sns.countplot(x='education', data=df, palette='coolwarm')
plt.title('Distribution of Educational Number')
plt.xlabel('Educational Number')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='race', hue='income', data=df, palette='crest')
plt.title('Income Distribution by Race')
plt.xlabel('Race')
plt.ylabel('Count')
plt.legend(title='Income', labels=['<=50K', '>50K'])
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python

plt.figure(figsize=(10, 6))
sns.histplot(df['fnlwgt'], kde=True, bins=50, color='red')
plt.title('Distribution of fnlwgt')
plt.xlabel('fnlwgt')
plt.ylabel('Count / Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python


plt.figure(figsize=(14, 7))
sns.countplot(x='occupation', data=df, palette='plasma')
plt.title('Distribution of Occupation')
plt.xlabel('Occupation')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='income', y='hours-per-week', data=df, palette='viridis')
plt.title('Hours per Week vs. Income')
plt.xlabel('Income')
plt.ylabel('Hours per Week')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
