```python
from abc import ABC, abstractmethod
class Account:
  def __init__(self,account_number,accountholder_name,initial_balance=0):
    self.account_number=account_number
    self.accountholder_name=accountholder_name
    self.balance=initial_balance
  @abstractmethod
  def deposit(self,amount):
    pass
  @abstractmethod
  def withdraw(self,amount):
    pass
  def show_details(self):
    print(f"Account Number: {self.account_number}")
    print(f"Account Holder Name: {self.accountholder_name}")
    print(f"Balance: {self.balance}")
class SavingsAccount(Account):

  def __init__(self, account_number, accountholder_name, balance, accountType):
        super().__init__(account_number, accountholder_name, balance)
        self.accountType = accountType

  def deposit(self,amount):
      self.balance+=amount
      print(f" Successfully deposited \n New Balance: {self.balance}")

  def withdraw(self,amount):
      if self.balance<amount:
        print("Insufficient balance")
      else:
        self.balance-=amount
        print(f"New Balance: {self.balance}")


  def show_details(self):
     print("----------------\n Account Details\n----------------------")
     super().show_details()
     print(f"Account Type: {self.accountType}")
class CurrentAccount(Account):
  def __init__(self, account_number, accountholder_name, balance, accountType):
        super().__init__(account_number, accountholder_name, balance)
        self.accountType = accountType
  def deposit(self,amount):
    self.balance+=amount
    print(f" Successfully deposited \n New Balance: {self.balance}")
  def withdraw(self,amount):
    self.balance-=amount
    print(f"New Balance: {self.balance}")

  def show_details(self):
     super().show_details()
     print(f"Account Type: {self.accountType}")

SavingAccount_list=[]
CurrentAccount_list=[]
accountholder={}
while(True):
  print("------------------------")
  print("1. New Account Creation")
  print("2. Deposit")
  print("3. Withdrawal")
  print("4. View Balance")
  print("5. Exit")
  print("------------------------")
  choice = int(input("Enter your choice: "))
  if choice ==1:
    account_number = input("Enter account number: ")
    account_holder_name=input("Enter account holder name: ")
    balance = int(input("Enter balance: "))
    accounttype=input("Enter account type: ")
    if accounttype=="Savings":
      accountholder = SavingsAccount(account_number,account_holder_name,balance,accounttype)
      SavingAccount_list.append(accountholder)
      for i in SavingAccount_list:
        i.show_details()
        break

    elif accounttype=="Current":
      accountholder = CurrentAccount(account_number,account_holder_name,balance,accounttype)
      CurrentAccount_list.append(accountholder)
      for i in CurrentAccount_list:
        i.show_details()
        break

  elif choice ==2:
    flag=False
    account_number = input("Enter account number: ")
    amount = int(input("Enter amount to deposit: "))
    for i in SavingAccount_list:
      if i.account_number == account_number:
        i.deposit(amount)
        flag=True
        break
    for i in CurrentAccount_list:
      if i.account_number == account_number:
        i.deposit(amount)
        flag=True
        break
    if(flag==False):
      print("Invalid Account Details")
  elif choice == 3:
      flag=False
      account_number = input("Enter account number: ")
      amount = int(input("Enter amount to withdraw: "))
      for i in SavingAccount_list:
        if i.account_number == account_number:
          i.withdraw(amount)
          flag=True
          break
      for i in CurrentAccount_list:
        if i.account_number == account_number:
          i.withdraw(amount)
          flag=True
          break
      if(flag==False):
        print("Invalid Account Details")
  elif choice ==4:
    flag=False
    account_number = input("Enter account number: ")
    for i in SavingAccount_list:
      if i.account_number == account_number:
        i.show_details()
        flag=True
        break
    for i in CurrentAccount_list:
      if i.account_number == account_number:
        i.show_details()
        flag=True
        break
    if(flag==False):
      print("Invalid Account Details")
  else:
    print("Thank you for visiting")
    break


```

```python

class Product:
  def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

  def display_info(self):
    print("---------------------")
    print(f"Name: {self.name}")
    print(f"Price: {self.price}")
    print(f"Quantity: {self.quantity}")
class Inventory:
    def __init__(self):
        self.products = []
    def addProduct(self, Product):
        self.products.append(Product)
    def updateProduct(self,product_name,price,quantity):
      for i in self.products:
        if i.name==product_name:
          i.price=price
          i.quantity=quantity
          break
    def deleteProduct(self,product_name):
      for i in self.products:
        if i.name==product_name:
          self.products.remove(i)
          break
    def display_info(self):
      print("**********************")
      for i in self.products:
        i.display_info()
    def total_inventory(self):
        total=0
        for i in self.products:
          total+=i.price
        print(f"Total Inventory: {total}")
    def __add__(self, other):
        combined_inventory = Inventory()
        combined_inventory.products = self.products + other.products
        return combined_inventory
pdt1=Product("Apple",100,10)
pdt2=Product("Banana",200,20)
pdt3=Product("Mango",300,30)
pdt4=Product("Orange",400,40)
inventory1=Inventory()
inventory2=Inventory()

inventory1.addProduct(pdt1)
inventory1.addProduct(pdt3)
inventory2.addProduct(pdt2)
inventory2.addProduct(pdt4)
inventory3=inventory1+inventory2
inventory1.display_info()
inventory1.total_inventory()
inventory2.display_info()
inventory2.total_inventory()
inventory1.updateProduct("Apple",300,30)
inventory1.display_info()
inventory1.deleteProduct("Mango")
inventory1.display_info()
inventory3=inventory1+inventory2
inventory3.display_info()
```
