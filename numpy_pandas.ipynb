```python
arr5 = np.random.randint(1, 100, size=36)
arr5 = arr5.reshape(6, 6)
print("Reshaped Array (6x6):")
print(arr5)

col_sum = np.sum(arr5, axis=0)
print("Column Sums:", col_sum)
row_sum = np.sum(arr5, axis=1)
print("Row Sums:", row_sum)
tot_sum = np.sum(arr5)
print("Total Sum:", tot_sum)
```

```python

stock_prices = np.array([100,102,98,105])
print("Stock Prices:")
print(stock_prices)
previous_prices = np.roll(stock_prices, 1)
print("Previous Prices:")
print(previous_prices)
daily_returns = (stock_prices - previous_prices) / previous_prices
print("Daily Returns:")
print(daily_returns)
# Set the first return to NaN since there's no previous price for the first day
daily_returns[0] = np.nan
print("Daily Returns:")
print(daily_returns)
```

```python

annual_rates = np.array([0.05, 0.06, 0.07])
monthly_rates = (1 + annual_rates) ** (1/12) - 1
print("Monthly Rates:")
print(monthly_rates)
```

```python
weights = np.array([0.2, 0.5, 0.3])
print("Weights:")
print(weights)
print("Sum of Weights:", np.sum(weights))
normalized_weights = weights / np.sum(weights)

# Print the normalized weights
print("Normalized Weights:")
print(normalized_weights)
print("Sum of Normalized Weights:", np.sum(normalized_weights))
```

```python

data1={
     "Account": [101,102,103,104],
    "Balance": [5000,15000,7000,12000]
}
# Create the DataFrame
df = pd.DataFrame(data1)
print(df)
# Filter rows where Balance > 8000
filtered_df1 = df[df["Balance"] > 8000]

# Filter rows where Balance is between 6000 and 13000
filtered_df2 = df[(df["Balance"] >= 6000) & (df["Balance"] <= 13000)]
print("Rows where Balance > 8000:")
print(filtered_df1)

print("\nRows where Balance is between 6000 and 13000:")
print(filtered_df2)

```
