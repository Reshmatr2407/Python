```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

filename='/content/Bengaluru_House_Data.csv'
df=pd.read_csv(filename)
df.head()
```

```python
df.shape
```

```python
df['society'].isna().sum()
```

```python
#percentage of missing values in a column
df.isna().sum()/len(df)*100
```

As the percentage of missing value in society are higher than 10% we eliminate that column

```python
df = df.drop(columns=['society'])
```

```python
df.columns
```

find outlier

```python
sns.boxplot(df)
```

Removing Outlier of bath

```python
Q1=df['bath'].quantile(0.25)
Q2=df['bath'].quantile(0.50)
Q3=df['bath'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
```

```python
df=df[(df['bath']>= lower) & (df['bath']<=upper)]
```
